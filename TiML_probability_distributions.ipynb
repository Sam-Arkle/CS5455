```python
# Generating the different probability distributions in Bishop's Pattern Recognition

# 1. Bernoulli distribution (coin toss distribution)
# We will generate this probability distribution using 'coin tosses'
import numpy as np
import matplotlib.pyplot as plt
N = 1000

coins = np.random.randint(0, 2, size=N)
heads = np.sum(coins)
tails = N - heads

labels = ['Head', 'Tails']
counts = [heads, tails]

plt.bar(labels, counts)
plt.xlabel('Coin face')
plt.ylabel('Frequency')
plt.title('Bernoulli Distribution')

plt.show()
```
