```python
from scipy.stats import bernoulli

p = 0.5
x = [0,1]
pmf_values = bernoulli.pmf(x, p)

plt.bar(x, pmf_values)

plt.xlabel('Event')
plt.ylabel('Probability')
plt.xticks(x)
plt.title('Bernoulli distribution with p=0.5')
plt.show()
```

```python
"""
The Beta distribution
This is a distribution over a continuous variable, often used to represent the probability for some binary event .
"""
from scipy.stats import beta
N = 100
x = np.linspace(0, 1, N)

ab_values = [(5, 5), (1, 1), (6, 2), (0.1, 0.1), (1.5, 3)]
colors = ['y', 'g', 'pink', 'r', 'b']
labels = [f'a={a}, b={b}' for a, b in ab_values]


for (a, b), color, label in zip(ab_values, colors, labels):
    plt.plot(x, beta.pdf(x, a, b), color=color, lw=2, label=label)

# Finalize the plot
plt.title('Beta distribution for different values of a and b')
plt.xlabel('X')
plt.ylabel('Probability Density')
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.show()
```

```python
# The binomial distribution
# The binomial distribution gives the probability of observing m occurrences of x = 1
# in a set of N samples from a Bernoulli distribution, where the probability of observing x = 1 is µ ∈ [0, 1].

from scipy.stats import binom

N = 100
probabilities = [0.75, 0.5, 0.05]
colors = ['red', 'blue', 'green']
labels = [f'p={p}' for p in probabilities]


x_values = list(range(N + 1))

for p, color, label in zip(probabilities, colors, labels):
    distribution = [binom.pmf(x, N, p) for x in x_values]
    plt.bar(x_values, distribution, color=color, label=label, alpha=0.6)

plt.legend()
plt.xlabel('X')
plt.ylabel('Probability')
plt.title('Binomial Distribution for Different Probabilities')
plt.show()


```

```python
# Gamma distribution
#The Gamma is a probability distribution over a positive random variable τ > 0
# governed by parameters a and b that are subject to the constraints a > 0 and b > 0
# to ensure that the distribution can be normalized

from scipy.stats import gamma

x = np.linspace(0,10,100)

alpha_beta_values = [(0.1,0.1),(2.8,1.4),(1,8)]
colours=['green','red','blue']
labels=[f'Gamma  distribution (α={alpha}, β={beta})' for alpha, beta in alpha_beta_values]

for (alpha, beta), colour, label in zip(alpha_beta_values, colours, labels):
  pdf = gamma.pdf(x, alpha, scale=beta)
  plt.plot(x, pdf, color=colour, label=label)


plt.title('Gamma Distribution for different alpha and beta values')
plt.xlabel('X')
plt.ylabel('Probability Density')
plt.legend()
plt.show()
```

```python
# The Gaussian (normal) distribution
from scipy.stats import norm

x = np.linspace(-5,5,1000) # We will use a mean of 0 and so we want the x values to be centered on 0

σ_values = [1,2,0.5] #standard deviations
colours=['blue','red','green']
labels = [f'σ = {σ}' for σ in σ_values]
for σ , colour, label in zip(σ_values, colours, labels):
  pdf = norm.pdf(x, loc=0, scale=σ)
  plt.plot(x, pdf, color=colour, label=label)

plt.title('Normal distribution for different values of σ')
plt.xlabel('X')
plt.ylabel('Probability Density')
plt.legend()
plt.show()
```

```python
# Uniform distribution

from scipy.stats import uniform

x = np.linspace(-10, 10 , 1000)
params = [(0, 1), (-0.5, 2), (-1.5, 4)]
colours = ['blue', 'red', 'green']
labels = [f'loc = {loc}, scale = {scale}' for loc, scale in params]

for (loc, scale), colour, label in zip(params, colours, labels):
    pdf = uniform.pdf(x, loc=loc, scale=scale)
    plt.plot(x, pdf, color=colour, label=label, alpha=0.6)

plt.title('Uniform Distribution for Different Parameters')
plt.xlabel('X')
plt.ylabel('Probability Density')
plt.legend()
plt.show()
```
